# Train CIFAR-100 → DiNN-style 3072 → 1024 → 100

Trains a bipolar MLP on CIFAR-100 **fine** labels and exports its weights as plain CSV files compatible with this repo's C++ FHE inference pipeline (see `cifar10/main.cpp` and `src/io/csv.cpp`).

Architecture: `3072 → 1024 → 100`

* **Input:** 32 × 32 × 3 RGB image, flattened in HWC order, with each byte thresholded at 127 to {-1, +1}.
* **Hidden:** `Linear(3072 → 1024)` followed by a hard-sign activation (straight-through estimator during training).
* **Output:** `Linear(1024 → 100)` — 100 fine-grained CIFAR-100 classes.

## 1. Setup and imports

````python
import os
import random
from pathlib import Path

import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
from torchvision import datasets

print("torch:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
````

## 2. Configuration

Everything tunable lives in `CONFIG`. The default `output_dir` is `cifar100/` (relative to the notebook's working directory) so the artifacts land in a sibling of the existing `cifar10/`-style folders when this notebook is run from the repo root or via `make up`.

````python
CONFIG = {
    "epochs": 30,
    "batch_size": 256,
    "lr": 1e-3,
    "weight_decay": 0.0,
    "device": "cuda" if torch.cuda.is_available() else "cpu",
    "seed": 0,
    "data_dir": "./data",
    "output_dir": "cifar100",
    "in_dim": 32 * 32 * 3,
    "hidden_dim": 1024,
    "num_classes": 100,
    # ── Quantization (matches DiNN-style integer weights used by cifar10/, MNIST_30/, MNIST_100/) ──
    # W1 is ternarized to {-1, 0, +1} using threshold τ = ternary_thresh * mean(|W1|)
    # (Ternary Weight Networks, Li & Liu 2016). W2 is rounded to signed integers
    # in [-w2_max_int, +w2_max_int]. Biases are rounded to the nearest integer.
    "ternary_thresh": 0.7,
    "w2_max_int":     15,
}


def seed_everything(seed: int) -> None:
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


seed_everything(CONFIG["seed"])
DEVICE = torch.device(CONFIG["device"])
print("Device:", DEVICE)
print("Output dir:", Path(CONFIG["output_dir"]).resolve())
````

## 3. CIFAR-100 dataset loading

We download CIFAR-100 with no transform so we keep the raw PIL images and can preprocess them ourselves — the same way the C++ side does, byte for byte.

`torchvision.datasets.CIFAR100` defaults to **fine** labels (0..99). Coarse (super-class) labels are not used here.

````python
train_pil = datasets.CIFAR100(
    root=CONFIG["data_dir"], train=True, download=True, transform=None
)
test_pil = datasets.CIFAR100(
    root=CONFIG["data_dir"], train=False, download=True, transform=None
)

print("Train:", len(train_pil), " Test:", len(test_pil))
print("Number of fine classes:", len(train_pil.classes))

_img, _label = train_pil[0]
print("Sample:", _img.size, _img.mode, "label =", _label, "(", train_pil.classes[_label], ")")
````

## 4. C++-compatible preprocessing

The C++ inference path does:

```cpp
// stb_image returns flat HWC interleaved RGB bytes when called with channels=3
unsigned char* data = stbi_load(path, &w, &h, &ch, /*channels=*/3);
for (int i = 0; i < 32*32*3; ++i) out[i] = (data[i] > 127) ? +1 : -1;
```

We replicate that exactly:

1. Convert PIL → NumPy `uint8` array of shape `(H, W, 3)` in **HWC** order.
2. Threshold each raw pixel byte at 127.
3. Map `> 127 → +1.0`, otherwise `-1.0`.
4. Flatten in HWC order (row, then column, then channel) to length 3072.

We deliberately **do not** use PyTorch's default `ToTensor` / CHW conversion — that would put all R bytes first, then all G, then all B, which is **not** the layout the C++ inference loop expects.

````python
def pil_to_bipolar_hwc(img) -> np.ndarray:
    """PIL → NumPy HWC uint8 → threshold at 127 → {-1, +1} float32, flat length 3072."""
    arr = np.asarray(img.convert("RGB"), dtype=np.uint8)  # (H, W, 3) HWC
    assert arr.shape == (32, 32, 3), f"unexpected shape {arr.shape}"
    bipolar = np.where(arr > 127, np.float32(1.0), np.float32(-1.0))
    flat = bipolar.reshape(-1)  # HWC row-major: (h, w, c) → single 3072-vector
    assert flat.shape == (3072,)
    return flat


def encode_dataset(pil_dataset) -> tuple[np.ndarray, np.ndarray]:
    n = len(pil_dataset)
    X = np.empty((n, 3072), dtype=np.float32)
    y = np.empty((n,), dtype=np.int64)
    for i in range(n):
        img, label = pil_dataset[i]
        X[i] = pil_to_bipolar_hwc(img)
        y[i] = label
    return X, y


X_train, y_train = encode_dataset(train_pil)
X_test, y_test = encode_dataset(test_pil)

print("X_train:", X_train.shape, X_train.dtype, "min/max =", X_train.min(), X_train.max())
print("X_test :", X_test.shape,  X_test.dtype,  "min/max =", X_test.min(),  X_test.max())
print("y_train classes:", len(np.unique(y_train)), "  y_test classes:", len(np.unique(y_test)))

# Sanity: every value is exactly +1 or -1.
assert set(np.unique(X_train)).issubset({-1.0, 1.0})
assert set(np.unique(X_test)).issubset({-1.0, 1.0})

train_ds = TensorDataset(torch.from_numpy(X_train), torch.from_numpy(y_train))
test_ds  = TensorDataset(torch.from_numpy(X_test),  torch.from_numpy(y_test))
train_loader = DataLoader(train_ds, batch_size=CONFIG["batch_size"], shuffle=True,  num_workers=0)
test_loader  = DataLoader(test_ds,  batch_size=CONFIG["batch_size"], shuffle=False, num_workers=0)
print("train batches:", len(train_loader), " test batches:", len(test_loader))
````

## 5. Model definition

DiNN-style MLP **with integer-only weights** (matches the format used by `cifar10/`, `MNIST_30/`, and `MNIST_100/` in this repo):

* `fc1 = Linear(3072, 1024)` — `W1` ternarized to `{-1, 0, +1}`, `b1` rounded to integer.
* hard-sign activation: `+1` if `x ≥ 0`, else `-1`.
* `fc2 = Linear(1024, 100)` — `W2` quantized to signed integers in `[-w2_max_int, +w2_max_int]`, `b2` rounded to integer.

Three straight-through estimators (STEs) make this trainable with regular SGD/Adam:

1. **`SignSTE`** for the hidden activation: forward is the true hard-sign, backward passes the gradient through, clipped to `|x| ≤ 1` (standard BinaryNet trick — keeps `fc1` weights from blowing up).
2. **`TernarizeSTE`** for `W1`: forward returns `sign(W) ⊙ 1[|W| ≥ τ]` with `τ = ternary_thresh · mean(|W|)` (Ternary Weight Networks, Li & Liu 2016); backward is the identity. The `Linear` layer keeps continuous-valued *shadow weights* that the optimizer updates; the integer values are derived from them on every forward.
3. **`IntQuantSTE`** for `W2`: forward scales `W` so its largest magnitude maps to `w2_max_int`, rounds to the nearest integer, and clamps to `[-w2_max_int, +w2_max_int]`; backward is the identity. Biases use a plain `RoundSTE`.

Because the **forward pass already operates on integer weights**, the training-time accuracy is exactly the accuracy you'll see when the exported CSVs are loaded back — no post-training quantization cliff.

````python
class SignSTE(torch.autograd.Function):
    """Forward = hard sign with +1 at zero. Backward = clamped identity (STE)."""

    @staticmethod
    def forward(ctx, x):
        ctx.save_for_backward(x)
        return torch.where(x >= 0, torch.ones_like(x), -torch.ones_like(x))

    @staticmethod
    def backward(ctx, grad_output):
        (x,) = ctx.saved_tensors
        # Pass gradient through, but kill it where |x| > 1 to keep fc1 weights tame.
        grad_input = grad_output * (x.abs() <= 1.0).to(grad_output.dtype)
        return grad_input


def sign_ste(x: torch.Tensor) -> torch.Tensor:
    return SignSTE.apply(x)


class TernarizeSTE(torch.autograd.Function):
    """Forward: w → sign(w) * 1[|w| ≥ τ], τ = thresh_factor * mean(|w|). Backward: identity."""

    @staticmethod
    def forward(ctx, w, thresh_factor):
        thresh = float(thresh_factor) * w.abs().mean()
        return torch.sign(w) * (w.abs() >= thresh).to(w.dtype)

    @staticmethod
    def backward(ctx, grad_output):
        return grad_output, None  # no gradient w.r.t. thresh_factor


def ternarize(w: torch.Tensor, thresh_factor: float = 0.7) -> torch.Tensor:
    return TernarizeSTE.apply(w, thresh_factor)


class IntQuantSTE(torch.autograd.Function):
    """Forward: round(w * (max_int / max(|w|))) clipped to ±max_int. Backward: identity."""

    @staticmethod
    def forward(ctx, w, max_int):
        w_max = w.abs().max().clamp(min=1e-8)
        scale = float(max_int) / w_max
        q = torch.round(w * scale).clamp(-float(max_int), float(max_int))
        return q

    @staticmethod
    def backward(ctx, grad_output):
        return grad_output, None  # no gradient w.r.t. max_int


def int_quant(w: torch.Tensor, max_int: int = 15) -> torch.Tensor:
    return IntQuantSTE.apply(w, max_int)


class RoundSTE(torch.autograd.Function):
    """Forward: round(x). Backward: identity. Used for biases."""

    @staticmethod
    def forward(ctx, x):
        return torch.round(x)

    @staticmethod
    def backward(ctx, grad_output):
        return grad_output


def round_ste(x: torch.Tensor) -> torch.Tensor:
    return RoundSTE.apply(x)


class DiNN(nn.Module):
    """3072 → hidden → num_classes with bipolar hidden activation and integer-only weights.

    Layer parameters are kept as continuous *shadow* tensors (so SGD/Adam can move
    them smoothly), but every forward pass derives integer weights from them via
    the STE quantizers above. That means `model(x)` is *exactly* what the exported
    CSVs produce — there is no accuracy gap between training-time and inference-time.
    """

    def __init__(self, in_dim: int, hidden_dim: int, num_classes: int,
                 ternary_thresh: float = 0.7, w2_max_int: int = 15):
        super().__init__()
        self.fc1 = nn.Linear(in_dim, hidden_dim)
        self.fc2 = nn.Linear(hidden_dim, num_classes)
        self.ternary_thresh = ternary_thresh
        self.w2_max_int     = w2_max_int

    def quantized_weights(self):
        """Return the integer weights actually used in the forward pass (and exported)."""
        W1 = ternarize(self.fc1.weight, self.ternary_thresh)
        b1 = round_ste(self.fc1.bias)
        W2 = int_quant(self.fc2.weight, self.w2_max_int)
        b2 = round_ste(self.fc2.bias)
        return W1, b1, W2, b2

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        W1, b1, W2, b2 = self.quantized_weights()
        h = torch.nn.functional.linear(x, W1, b1)
        h = sign_ste(h)
        return torch.nn.functional.linear(h, W2, b2)


model = DiNN(
    CONFIG["in_dim"], CONFIG["hidden_dim"], CONFIG["num_classes"],
    ternary_thresh=CONFIG["ternary_thresh"],
    w2_max_int=CONFIG["w2_max_int"],
).to(DEVICE)
n_params = sum(p.numel() for p in model.parameters())
print(model)
print("Trainable params:", n_params)
print(f"W1 quantization: ternary {{-1, 0, +1}}, threshold = {CONFIG['ternary_thresh']} * mean(|W1|)")
print(f"W2 quantization: signed integers in [-{CONFIG['w2_max_int']}, +{CONFIG['w2_max_int']}]")
````

## 6. Training loop

````python
optimizer = torch.optim.Adam(
    model.parameters(), lr=CONFIG["lr"], weight_decay=CONFIG["weight_decay"]
)
criterion = nn.CrossEntropyLoss()


def evaluate(model, loader, device):
    model.eval()
    correct = 0
    total = 0
    loss_sum = 0.0
    with torch.no_grad():
        for xb, yb in loader:
            xb = xb.to(device)
            yb = yb.to(device)
            logits = model(xb)
            loss = criterion(logits, yb)
            loss_sum += loss.item() * xb.size(0)
            pred = logits.argmax(dim=1)
            correct += (pred == yb).sum().item()
            total += xb.size(0)
    return loss_sum / total, correct / total


# Per-epoch history so we can plot loss / accuracy curves later.
history = {
    "epoch":      [],
    "train_loss": [],
    "train_acc":  [],
    "test_loss":  [],
    "test_acc":   [],
}

for epoch in range(1, CONFIG["epochs"] + 1):
    model.train()
    epoch_loss = 0.0
    epoch_correct = 0
    epoch_n = 0
    for xb, yb in train_loader:
        xb = xb.to(DEVICE)
        yb = yb.to(DEVICE)
        optimizer.zero_grad()
        logits = model(xb)
        loss = criterion(logits, yb)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item() * xb.size(0)
        epoch_correct += (logits.argmax(dim=1) == yb).sum().item()
        epoch_n += xb.size(0)

    train_loss = epoch_loss / epoch_n
    train_acc  = epoch_correct / epoch_n
    test_loss, test_acc = evaluate(model, test_loader, DEVICE)

    history["epoch"].append(epoch)
    history["train_loss"].append(train_loss)
    history["train_acc"].append(train_acc)
    history["test_loss"].append(test_loss)
    history["test_acc"].append(test_acc)

    print(
        f"epoch {epoch:3d}/{CONFIG['epochs']}  "
        f"train_loss={train_loss:.4f}  train_acc={train_acc*100:.2f}%  "
        f"test_loss={test_loss:.4f}  test_acc={test_acc*100:.2f}%"
    )
````

## 6b. Training curves

Two side-by-side plots from the per-epoch `history` collected above:

* **Loss vs epoch** — train loss (running average over the epoch) and test loss.
* **Accuracy vs epoch** — train accuracy (computed on the live, mid-training predictions) and test accuracy.

The train-loss curve is a *running* average across the epoch's mini-batches, so it lags slightly behind the true end-of-epoch value — that's expected and is the standard way to plot it cheaply.

````python
import matplotlib.pyplot as plt

epochs = history["epoch"]

fig, (ax_loss, ax_acc) = plt.subplots(1, 2, figsize=(12, 4.5))

ax_loss.plot(epochs, history["train_loss"], label="train loss", marker="o", markersize=3)
ax_loss.plot(epochs, history["test_loss"],  label="test loss",  marker="o", markersize=3)
ax_loss.set_xlabel("epoch")
ax_loss.set_ylabel("cross-entropy loss")
ax_loss.set_title("CIFAR-100 DiNN: loss vs epoch")
ax_loss.grid(True, alpha=0.3)
ax_loss.legend()

ax_acc.plot(epochs, [a * 100 for a in history["train_acc"]], label="train acc", marker="o", markersize=3)
ax_acc.plot(epochs, [a * 100 for a in history["test_acc"]],  label="test acc",  marker="o", markersize=3)
ax_acc.set_xlabel("epoch")
ax_acc.set_ylabel("accuracy (%)")
ax_acc.set_title("CIFAR-100 DiNN: accuracy vs epoch")
ax_acc.grid(True, alpha=0.3)
ax_acc.legend()

best_epoch = int(np.argmax(history["test_acc"])) + 1
best_acc   = max(history["test_acc"]) * 100
print(f"Best test accuracy: {best_acc:.2f}% at epoch {best_epoch} (chance = 1.00%).")

fig.tight_layout()
plt.show()
````

## 7. Evaluation

````python
final_train_loss, final_train_acc = evaluate(model, train_loader, DEVICE)
final_test_loss,  final_test_acc  = evaluate(model, test_loader,  DEVICE)
print(f"Final train accuracy: {final_train_acc*100:.2f}%  (loss {final_train_loss:.4f})")
print(f"Final test  accuracy: {final_test_acc*100:.2f}%  (loss {final_test_loss:.4f})")
````

## 8. CSV export

We export the **integer** weights actually used by the forward pass (no further quantization or rescaling), no headers, with `fmt="%d"`. This matches the file convention already used by `cifar10/`, `MNIST_30/`, and `MNIST_100/` in this repo.

| File | Shape | Value range |
|---|---|---|
| `cifar100_weights_W1.csv` | `(1024, 3072)` | `{-1, 0, +1}` (ternary) |
| `cifar100_weights_b1.csv` | `(1024,)`      | small signed ints |
| `cifar100_weights_W2.csv` | `(100, 1024)`  | signed ints in `[-w2_max_int, +w2_max_int]` |
| `cifar100_weights_b2.csv` | `(100,)`       | small signed ints |

`torch.nn.Linear.weight` is already stored as `(out_features, in_features)`, which matches the C++ `LoadCsv2D(path, expected_in, expected_out)` preferred orientation, so we can dump the parameters directly without transposing.

````python
output_dir = Path(CONFIG["output_dir"])
output_dir.mkdir(parents=True, exist_ok=True)

# Pull the *quantized* (integer) tensors actually used by the forward pass.
model.eval()
with torch.no_grad():
    W1_t, b1_t, W2_t, b2_t = model.quantized_weights()

W1 = W1_t.detach().cpu().numpy().astype(np.int64)  # (hidden_dim, 3072), values in {-1, 0, +1}
b1 = b1_t.detach().cpu().numpy().astype(np.int64)  # (hidden_dim,)
W2 = W2_t.detach().cpu().numpy().astype(np.int64)  # (num_classes, hidden_dim), values in [-w2_max_int, +w2_max_int]
b2 = b2_t.detach().cpu().numpy().astype(np.int64)  # (num_classes,)

# Sanity: every value really is an integer, with the expected range.
assert set(np.unique(W1).tolist()).issubset({-1, 0, 1}), \
    f"W1 not ternary: unique={np.unique(W1)}"
assert int(np.abs(W2).max()) <= CONFIG["w2_max_int"], \
    f"W2 outside ±{CONFIG['w2_max_int']}: max(|W2|)={int(np.abs(W2).max())}"

print("W1 unique values:", np.unique(W1).tolist())
print(f"b1 range: [{int(b1.min())}, {int(b1.max())}]")
print(f"W2 range: [{int(W2.min())}, {int(W2.max())}]   max(|W2|) = {int(np.abs(W2).max())}")
print(f"b2 range: [{int(b2.min())}, {int(b2.max())}]")

paths = {
    "W1": output_dir / "cifar100_weights_W1.csv",
    "b1": output_dir / "cifar100_weights_b1.csv",
    "W2": output_dir / "cifar100_weights_W2.csv",
    "b2": output_dir / "cifar100_weights_b2.csv",
}

# 2D matrices: comma-separated rows, no headers. 1D vectors: one value per line.
# fmt="%d" matches cifar10/MNIST_30/MNIST_100 — the FHE pipeline expects integer weights.
np.savetxt(paths["W1"], W1, delimiter=",", fmt="%d")
np.savetxt(paths["W2"], W2, delimiter=",", fmt="%d")
np.savetxt(paths["b1"], b1, fmt="%d")
np.savetxt(paths["b2"], b2, fmt="%d")

print()
for name, p in paths.items():
    print(f"{name}: {p}  ({p.stat().st_size} bytes)")
````

## 9. Export verification

Reload the CSVs with NumPy and check shapes **and exact integer parity** (since we wrote `%d`, the round-trip should be lossless: `max |W − W_reloaded| == 0`).

````python
W1_rl = np.loadtxt(paths["W1"], delimiter=",", dtype=np.int64)
b1_rl = np.loadtxt(paths["b1"],                dtype=np.int64)
W2_rl = np.loadtxt(paths["W2"], delimiter=",", dtype=np.int64)
b2_rl = np.loadtxt(paths["b2"],                dtype=np.int64)

H, K, IN = CONFIG["hidden_dim"], CONFIG["num_classes"], CONFIG["in_dim"]
assert W1_rl.shape == (H,  IN), W1_rl.shape
assert b1_rl.shape == (H,),     b1_rl.shape
assert W2_rl.shape == (K,  H),  W2_rl.shape
assert b2_rl.shape == (K,),     b2_rl.shape

# Integer round-trip: every diff must be exactly zero.
assert np.array_equal(W1, W1_rl), "W1 round-trip mismatch"
assert np.array_equal(b1, b1_rl), "b1 round-trip mismatch"
assert np.array_equal(W2, W2_rl), "W2 round-trip mismatch"
assert np.array_equal(b2, b2_rl), "b2 round-trip mismatch"

print("max |W1 - W1_rl|:", int(np.max(np.abs(W1 - W1_rl))))
print("max |b1 - b1_rl|:", int(np.max(np.abs(b1 - b1_rl))))
print("max |W2 - W2_rl|:", int(np.max(np.abs(W2 - W2_rl))))
print("max |b2 - b2_rl|:", int(np.max(np.abs(b2 - b2_rl))))
print("Shapes and integer values OK.")
````

## 10. Exported-model sanity check

Run inference using **the reloaded CSVs** with the exact bipolar pipeline that the C++ side will use:

```cpp
hidden[j] = (b1[j] + sum_i W1[j][i] * pixels[i] >= 0) ? +1 : -1;
scores[j] = b2[j] + sum_i W2[j][i] * hidden[i];
predicted = argmax(scores);
```

Then compare against the live PyTorch model. Since the weights are now **exact integers** and the round-trip is lossless (no `%.8f` rounding), prediction agreement should be exactly **100%** and accuracy should match between the two paths bit-for-bit.

````python
def numpy_inference(X: np.ndarray, W1, b1, W2, b2) -> np.ndarray:
    """Pure-integer forward pass, mirroring the C++ side bit-for-bit."""
    # Promote to int64 so the ~3072-term dot products can't overflow.
    X64  = X.astype(np.int64)
    W1_64 = W1.astype(np.int64)
    b1_64 = b1.astype(np.int64)
    W2_64 = W2.astype(np.int64)
    b2_64 = b2.astype(np.int64)

    h_pre  = X64 @ W1_64.T + b1_64                     # (N, 100), int64
    h      = np.where(h_pre >= 0, 1, -1).astype(np.int64)  # bipolar hard-sign
    scores = h @ W2_64.T + b2_64                       # (N, 100), int64
    return np.argmax(scores, axis=1)


# X_test is float32 with values in {-1.0, +1.0}; cast to int for the integer path.
csv_pred = numpy_inference(X_test.astype(np.int64), W1_rl, b1_rl, W2_rl, b2_rl)
csv_acc  = float((csv_pred == y_test).mean())

model.eval()
torch_pred_chunks = []
with torch.no_grad():
    for xb, _ in test_loader:
        xb = xb.to(DEVICE)
        logits = model(xb)
        torch_pred_chunks.append(logits.argmax(dim=1).cpu().numpy())
torch_pred = np.concatenate(torch_pred_chunks)
torch_acc  = float((torch_pred == y_test).mean())

agreement = float((csv_pred == torch_pred).mean())

print(f"PyTorch test accuracy:    {torch_acc*100:.2f}%")
print(f"Reloaded CSV accuracy:    {csv_acc*100:.2f}%")
print(f"Prediction agreement:     {agreement*100:.2f}%   (expected: 100.00% — pure-integer round-trip)")
````

## 11. Summary

**Where the CSVs were written:** `CONFIG["output_dir"]` (default `cifar100/`), as four headerless **integer** files (`fmt="%d"`, same convention as `cifar10/`, `MNIST_30/`, `MNIST_100/`):

| File | Shape | Source | Value range |
|---|---|---|---|
| `cifar100_weights_W1.csv` | `1024 × 3072` | `model.quantized_weights()[0]` (ternarized `fc1.weight`) | `{-1, 0, +1}` |
| `cifar100_weights_b1.csv` | `1024`        | `model.quantized_weights()[1]` (rounded `fc1.bias`)    | small signed ints |
| `cifar100_weights_W2.csv` | `100 × 1024`  | `model.quantized_weights()[2]` (int-quantized `fc2.weight`) | `[-w2_max_int, +w2_max_int]` |
| `cifar100_weights_b2.csv` | `100`         | `model.quantized_weights()[3]` (rounded `fc2.bias`)    | small signed ints |

**How these map to a future `cifar100/main.cpp`:** mirror `cifar10/main.cpp` and just change three constants plus the weight paths:

```cpp
static constexpr int IN_DIM  = 3072;  // 32 * 32 * 3, same as CIFAR-10
static constexpr int HID_DIM = 1024;  // was 30 in cifar10
static constexpr int OUT_DIM = 100;   // was 10  in cifar10  (CIFAR-100 has 100 classes)

auto W1 = io::LoadCsv2D("../cifar100_weights_W1.csv", IN_DIM,  HID_DIM);
auto b1 = io::LoadCsv1D("../cifar100_weights_b1.csv");
auto W2 = io::LoadCsv2D("../cifar100_weights_W2.csv", HID_DIM, OUT_DIM);
auto b2 = io::LoadCsv1D("../cifar100_weights_b2.csv");

auto pixels = io::LoadImageBipolar(argv[1], IN_DIM, /*channels=*/3);
```

Because we exported the weights as `(out, in)` (the C++ preferred orientation), `LoadCsv2D` will accept them without needing to transpose.

> **Naming reminder.** `CIFAR-100` here means **100 output classes** (the hidden-layer width is a separate hyperparameter — currently 1024). That's different from `MNIST_100`, where the `100` refers to the **hidden-layer width** and there are still only 10 output classes.
